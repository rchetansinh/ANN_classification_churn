```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
## load the data set
data =pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
## Proprocess the data
### Drop irrelevant columns
data =data.drop(["RowNumber","CustomerId","Surname"],axis=1) ##axis = 1 is columns horizontally 0 means row verically
data
```

```python
## Encode catogorical veriables

label_encoder_gender = LabelEncoder()
data["Gender"] = label_encoder_gender.fit_transform(data["Gender"])
data
```

```python
#OHE gepgraphy cols

from sklearn.preprocessing import OneHotEncoder
onehot_encoder_geo= OneHotEncoder()
geo_encoder = onehot_encoder_geo.fit_transform(data[["Geography"]])
geo_encoder
```

```python
geography_fetures = onehot_encoder_geo.get_feature_names_out(["Geography"])
geography_fetures
```

```python
#geo_encoded_df= pd.DataFrame(geo_encoder,columns=geography_fetures)
#geo_encoded_df
```

```python
#fix issue converting in toarray()
geo_encoded_df= pd.DataFrame(geo_encoder.toarray(),columns=geography_fetures)
geo_encoded_df.head()
```

```python
### Combine one hot encoder columns with original data

#data = data.combine(geo_encoded_df)
#data = data.drop(data.Geography,axis=1)
data = pd.concat([data,geo_encoded_df],axis=1)
data =data.drop("Geography",axis=1)
data
```

```python
#Save the encoder and sscaler

with open("label_encoder_gender.pkl","wb") as file:
    pickle.dump(label_encoder_gender,file)

with open("onehot_encoder_geo.pkl","wb") as file:
    pickle.dump(onehot_encoder_geo,file)

```

```python
## devide the dataset into indepent and dependent features
data.head()
x = data.drop("Exited",axis=1) ##dependent veribale
y = data.Exited # indipent veriable

##split data into traning and testing sets

X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=.2,random_state=42)


# scale there features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```python
with open("scaler.pkl","wb") as file:
    pickle.dump(scaler,file)
```

```python
##ANN Implementation 
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
X_train.shape
```

```python
## Build Our ANN Model
model =Sequential([
    Dense(64,activation="relu",input_shape=(X_train.shape[1],)), # HL 1 connected with i/p layer
    Dense(32,activation="relu"),  #HL2
    Dense(1,activation="sigmoid") # output layer
    
])
```

```python
model.summary()
```

```python
import tensorflow
opt =tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
loss = tensorflow.keras.losses.BinaryCrossentropy()
loss
```

```python
##. compile the model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
## optimizer if we using like this than it will take automatocally the learning rate if
## if we want to change it we need to use as above
```

```python
### set up the Tensorboard
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard

log_dir = "logs/fit" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback =TensorBoard(
    log_dir=log_dir,
    histogram_freq=1
)
```

```python
##Set up Early Stopping

early_stopping_callback =EarlyStopping(
    monitor="val_loss",
    patience= 10, ##wait till 5 epoch
    restore_best_weights=True
)
```

```python
### Train the Model
history =model.fit(
    X_train,
    Y_train,
    validation_data=(X_test,Y_test),
    epochs=100,
    callbacks=[tensorflow_callback,early_stopping_callback]
)
```

```python
model.save("model.h5")
```
