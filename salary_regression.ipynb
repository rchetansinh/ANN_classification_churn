```python
##from same data set will select estimate salary as independent

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler
```

```python
data = pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
## preprocessing the data

data = data.drop(["RowNumber" , "CustomerId" , "Surname"],axis=1)
data.head()
```

```python
# label encoder
label_encoder_gender = LabelEncoder()
data.Gender = label_encoder_gender.fit_transform(data.Gender)
data.head()
```

```python
# One hot encoder

onehot_encoder_geo = OneHotEncoder(handle_unknown="ignore")
geo_encoder =onehot_encoder_geo.fit_transform(data[["Geography"]]).toarray()
geo_encoder_df =pd.DataFrame(geo_encoder,columns=onehot_encoder_geo.get_feature_names_out(["Geography"]))
geo_encoder_df
```

```python
# Combine one-hot encoded columns with original data
data = pd.concat([data.drop("Geography",axis=1),geo_encoder_df],axis=1)
data 
```

```python
# Split the data into features and target

x =data.drop("EstimatedSalary",axis=1) #indiependent veriable
print(x)
y = data.EstimatedSalary #dependent veriable 
print(y)
```

```python
## Splite data into training and testing sets

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
## scale the feature

scaler = StandardScaler()
X_train =scaler.fit_transform(x_train)
X_test =scaler.fit_transform(x_test)
```

```python
#save encoder and scaler for later user
import pickle 
with open("label_encoder_gender_regression.pkl","wb") as file:
    pickle.dump(label_encoder_gender,file)

with open("onehot_encoder_geo_regression.pkl","wb") as file:
    pickle.dump(onehot_encoder_geo,file)

with open("scaler_regression.pkl","wb") as file:
    pickle.dump(scaler,file)
```

## ANN regression problem statement 

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X_train.shape
```

```python
## Build model

model = Sequential([
    Dense(64,activation="relu", input_shape = (X_train.shape[1],)),
    Dense(32,activation="relu"),
    Dense(1)
])

##compile the model
model.compile(optimizer = 'adam',loss='mean_absolute_error',metrics=['mae'])

model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime

# set up tensorboard

log_dir = "regressionlogs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
# Set up Early Stopping
early_stopping_callback = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
# Train the model
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=100,
    callbacks=[early_stopping_callback, tensorboard_callback]
)
```

```python
model.save('regression_model.h5')
```

```python
### Evaluate model on test data

test_loss,test_mae = model.evaluate(X_test,y_test)
print(f"test MAE : {test_mae}")
```
